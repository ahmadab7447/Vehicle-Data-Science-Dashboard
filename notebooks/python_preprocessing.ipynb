```python
import pandas as pd
import numpy as np

file_path = "VEXAR_Fleet_Dataset_CANDIDATE_VERSION (2).xlsx"

drivers = pd.read_excel(
    file_path,
    sheet_name="Drivers",
    header=1
)

vehicles = pd.read_excel(
    file_path,
    sheet_name="Vehicles",
    header=1
)

trips = pd.read_excel(
    file_path,
    sheet_name="Trips",
    header=1
)

telemetry = pd.read_excel(
    file_path,
    sheet_name="Telemetry",
    header=1
)
```

```python
def fix_headers(df):
    df.columns = df.iloc[0]
    df = df.iloc[1:].reset_index(drop=True)
    return df

drivers = fix_headers(drivers)
vehicles = fix_headers(vehicles)
trips = fix_headers(trips)
telemetry = fix_headers(telemetry)
```

```python
driver_numeric = [
    "Age",
    "License_Experience_Years"
]

vehicle_numeric = [
    "Manufacture_Year",
    "Odometer_KM_Start_of_Week"
]

trip_numeric = [
    "Duration_Min",
    "Distance_KM",
    "Avg_Speed_kmph",
    "Max_Speed_kmph",
    "Start_Latitude",
    "Start_Longitude",
    "End_Latitude",
    "End_Longitude"
]

telemetry_numeric = [
    "Latitude",
    "Longitude",
    "Speed_kmph",
    "Accel_X_g",
    "Accel_Y_g",
    "Accel_Z_g",
    "Gyro_X_dps",
    "Gyro_Y_dps",
    "Gyro_Z_dps"
]
```

```python
drivers["Date_Joined_Fleet"] = pd.to_datetime(
    drivers["Date_Joined_Fleet"],
    errors="coerce"
)
```

```python
vehicles["Registration_Date"] = pd.to_datetime(
    vehicles["Registration_Date"],
    errors="coerce"
)

vehicles["Last_Service_Date"] = pd.to_datetime(
    vehicles["Last_Service_Date"],
    errors="coerce"
)
```

```python
trips["Trip_Date"] = pd.to_datetime(
    trips["Trip_Date"],
    errors="coerce"
)



```

```python
trips["Start_DateTime"] = pd.to_datetime(
    trips["Trip_Date"].dt.strftime("%Y-%m-%d") + " " +
    trips["Start_Time"].astype(str)
)

trips["End_DateTime"] = pd.to_datetime(
    trips["Trip_Date"].dt.strftime("%Y-%m-%d") + " " +
    trips["End_Time"].astype(str)
)
```

```python
telemetry["Timestamp"] = pd.to_datetime(
    telemetry["Timestamp"],
    errors="coerce"
)
```

```python
telemetry = telemetry.sort_values(
    ["Trip_ID", "Timestamp"]
).reset_index(drop=True)
```

```python
telemetry["Time_Diff_Min"] = (
    telemetry.groupby("Trip_ID")["Timestamp"]
    .diff()
    .dt.total_seconds() / 60
)
```

```python
telemetry_counts = (
    telemetry.groupby("Trip_ID")
    .size()
    .reset_index(name="Telemetry_Count")
)

trip_check = trips.merge(
    telemetry_counts,
    on="Trip_ID",
    how="left"
)

trip_check[
    ["Trip_ID", "Duration_Min", "Telemetry_Count"]
].head(10)
```
